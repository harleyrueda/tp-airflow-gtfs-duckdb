```python
import duckdb
import pandas as pd
from datetime import datetime
from zoneinfo import ZoneInfo
import os
```

```python
con = duckdb.connect()
```

```python
con.execute("""
    CREATE OR REPLACE VIEW delays_csv AS
    SELECT *
    FROM read_csv_auto('./exports/avg_delay_by_minute_*.csv', header=True)
""")
```

```python
df = con.execute("""
    SELECT
        strftime(minute_ts, '%Y-%m-%d') AS jour_date,
        strftime(minute_ts, '%H') AS heure,
        ROUND(AVG(avg_delay_min), 2) AS avg_delay_min,
        COUNT(*) AS nb_events
    FROM delays_csv
    WHERE minute_ts IS NOT NULL
    AND minute_ts <= now()
    GROUP BY jour_date, heure
    ORDER BY jour_date, heure
""").df()

df
```
